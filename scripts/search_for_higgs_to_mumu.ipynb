#### Fermions

Each "lepton" has an electrically neutral counterpart, called neutrinos, namely, electron neutrino, muon neutrino, and tau neutrino.  Quarks consist of six flavours, namely, up, down, charm, strange, bottom (or beauty), and top. There are three up-type quarks, namely, up, charm, and top, and they are so named because they have positive electric charge. Additionally, the strong nuclear force says that free quarks and gluons combine to form composite particles called hadrons, also called hadronization. This hadronization time occurs within 10^-23 seconds. 

All the fundamental particles in the SM have an anti-particle counterpart which is identical in every way except that they posses the opposite electrical charge.

#### Particle Interaction

According to the SM, there are only few types of interactions that are allowed. All electrically charged particles can interact via the photon. Since the photon itself is electrically neutral, it does not interact with itself. All colour charged particles can interact via the gluon. Since the gluon itself is colour charged, it can interact with itself. All fermions possess the weak charge and therefore can interact via the W or Z bosons. Since the W boson carries the weak charge, it can interact with itself. The Z boson does not carry the weak charge, so it cannot interact with itself. 

The photon and Z boson has no structural impact on the particle, e.g., an electron to an electron. The gluon has no structural impact on the particle, but does change its colour charge, e.g., green top quark to anti-blue top quark. The W boson is the only boson interaction that changes the flavour of the particle. The W boson changes an up-type quark into a down-type quark, or vice versa, e.g., an top quark into a bottom quark. Additionally, the W boson changes a lepton into its corresponding lepton-neutrino e.g., electron to electron neutrino.

Furthermore,photons and gluons can undergo pair production, where the photon produces a lepton and its corresponding anti-lepton, while the gluon produces a quark and its corresponding anti-quark.

### The Higgs Boson

The Higgs boson has an electric charge of 0, spin of 0, and a mass of approximately 125 GeV. It is a manifestation of the Higgs field, which gives mass to the fundamental particles.

#### The Higgs Field

In quantum field theory, every fundamental particle is an excitation of its respective pervasive field (like photons in the electromagnetic field). The Higgs field is a unique scalar field that permeates all of spacetime, but unlike other fields, it possesses a non-zero vacuum expectation value in empty space. Fundamental particles like quarks and electrons acquire their mass through their continuous interactions with this non-zero Higgs field. The stronger a particle couples to the Higgs field, the more massive it becomes, while particles that do not interact with it (such as photons) remain massless. Finally, the Higgs boson is simply a quantized fluctuation, or localized excitation, of the Higgs field itself.

#### Higgs Boson Production

Based on the allowed interactions of the SM, the Higgs boson can be produced in a variety of ways. However, due to its heavy mass and specific couplings, its production is highly dependent on the center-of-mass energy and the type of colliding particles.

In electron-positron colliders, the Higgs boson cannot easily be produced alone via annihilation because its coupling to light leptons is proportional to their tiny masses. Instead, it is predominantly produced at high energies via "Higgsstrahlung," where an electron and positron annihilate into a virtual $Z$ boson that radiates a Higgs boson. Because circular electron accelerators suffer heavily from synchrotron radiation—which is inversely proportional to the fourth power of the particle's mass—achieving the high energies required for this process is difficult, making massive proton-proton colliders the preferred choice for high-rate production.

In hadron colliders like the LHC, protons provide the necessary high energy with minimal synchrotron loss, allowing constituents (partons) to drive Higgs production. The dominant production mode is gluon-gluon fusion, which accounts for nearly 90% of the total cross-section. Because gluons are massless and cannot couple to the Higgs directly, this process occurs through a quantum loop of virtual heavy quarks, dominated by the top quark due to its enormous coupling to the Higgs (known as the Yukawa coupling). Other significant hadronic channels include vector boson fusion, where quarks radiate $W$ or $Z$ bosons that fuse into a Higgs, and associated production with a vector boson or a top-antitop quark pair.

#### Higgs Boson Decay

The Higgs has a very short lifetime, on the order of $10^{-22}~\text{s}$. It decays extremely quickly after production, so there is no hope of directly detecting the particle. 

The Higgs boson has a wide variety of decay channels because it couples directly to the mass of Standard Model particles. Unlike the top quark, it does not have a single dominant decay mode that approaches 100%, though it most frequently decays into a bottom-antibottom quark pair: $H\rightarrow b\bar{b}$. This channel accounts for approximately 58% of all Higgs decays. 

Because the Higgs boson's mass is around 125 GeV, it cannot decay into a real top-antitop pair. However, it can decay into pairs of the heavy electroweak gauge bosons, specifically $W^{+}W^{-}$ (approx. 21.5%) and $ZZ$ (approx. 2.6%). Because a Higgs boson is lighter than two on-shell gauge bosons, one or both of these bosons must be virtual ($W^{*}$ or $Z^{*}$).

The remaining decay branching ratios include decays into tau leptons ($H \rightarrow \tau^+\tau^-$, ~6.3%), gluons ($H \rightarrow gg$, ~8.2% via a top quark loop), and photons ($H \rightarrow \gamma\gamma$, ~0.2% via $W$ boson and top quark loops). While the photon channel is rare, its clean experimental signature makes it one of the primary discovery channels. 

Additionally, the Higgs can decay into second-generation fermions, such as a muon-antimuon pair ($H \rightarrow \mu^+\mu^-$). Because muons are significantly lighter than third-generation tau leptons, this decay is exceptionally scarce, occurring in only about 0.02% of all Higgs decays (approximately 1 in 5,000 events).

#### Higgs Boson Importance

The Higgs boson is the only fundamental scalar (spin-0) particle in the SM, making it entirely unique compared to the spin-1/2 fermions and spin-1 gauge bosons. Because it is the physical manifestation of the Higgs mechanism, studying its properties allows for a direct probe into the electroweak symmetry breaking sector and map out the shape of the Higgs potential, which determines the stability of our universe. 

The Higgs boson has a mass of approximately 125 GeV, giving it a theoretical lifetime of about $10^{-22}$ seconds. This lifetime ensures that the Higgs cannot be observed directly before it decays, meaning its properties must be meticulously reconstructed from its diverse decay products. Precision measurements of its coupling strengths to other particles are critical; since the Higgs couples directly to mass, any slight deviation from the predicted SM coupling values would serve as a clear smoking gun for new physics.

Furthermore, the Higgs boson is centrally linked to the "hierarchy problem"—the question of why the Higgs mass is so light despite receiving enormous quantum corrections from high-energy scales. This puzzle makes the Higgs a central focus for many Beyond the Standard Model (BSM) theories, such as Supersymmetry or Composite Higgs models, which predict new partner particles that cancel out these quantum corrections.

### The Large Hadron Collider

The Large Hadron Collider (LHC) is a 27 km circular ring that spans across France and Switzerland. It consists of four major experiments with detectors situated along the ring. These four experiments are ATLAS, CMS, ALICE, and LHCb. The ATLAS and CMS detectors were designed as general purpose detectors to perform a wide variety of studies. ALICE, however, was designed to study heavy-ion collisions and the quark-gluon plasma, while LHCb was designed to study heavy-flavour hadrons, specifically bottom quarks - hence the "b" in the name. 

The large hadron collider is the largest circular collider in the world and is the last accelerator in the chain before the protons collide. A hydrogen gas bottle releases hydrogen atoms. An electric field is used to knock out the electron from the hydrogen atom, which leaves behind the proton. These protons are fed into multiple accelerators to speed them up to nearly the speed of light with energies of 6.5 TeV. These protons are bunched up together before colliding at the centre of the four major experiments. The aftermath of these collisions are then captured by the detectors and the data is stored for further analysis.

### The ATLAS Detector

The ATLAS detector is the largest detector in the world and is cylindrical in shape, which is about 45 metres long and 25 metres in diameter. Many subdetectors make the ATLAS detector starting from the Inner Detector, which is the closest subdetector to the collision point, all the way to the Muon Spectrometer, which is the outermost subdetector. Additionally, large magnets are used to focus the proton bunches and to help identify particles that come from the collision.

The Inner Detector (ID) is used to measure the trajectory of the particles that come from the collisions. These trajectories are reconstructed from the signals that are left in the detector as the particles interact with the material of the sensors. The next subdetector is the Electric Calorimeter (ECal) which is used to measure the energy of electromagnetically particles, specifically, electrons and photons. The Hadronic Calorimeter (HCal) follows and is used to measure the energy of hadrons. The last subdetector is the Muon Spectrometer (MS) which is used to measure the tracks and energy of muons.

#### Number of events

The number of events observed in an experiment depends on two factors:
1. How frequently the process occurs.
2. How much data have been collected.

The amount of data collected is quantified using the **integrated luminosity**. A larger integrated luminosity corresponds to more proton-proton collisions and therefore more opportunities for rare processes to occur.

Therefore, to determine the number of simulation events we should theoretically produce, we use 

$$N = \mathcal{L}\sigma$$

where $N$ is the number of events, $\mathcal{L}$ is the integrated luminosity, and $\sigma$ is the cross-section of that specific process.

##### Event weights

Imagine that two simulation samples each contain 100,000 events. Does this mean the corresponding physical processes occur equally often in nature? Not necessarily. One process may occur much more frequently than the other. The number of simulated events is chosen by physicists and does not directly reflect the true production rate of the process. Event weights allow us to convert simulation samples into realistic predictions.

In a full particle physics analysis, event weights include contributions from:

- production values (cross-sections, k-factors, filter efficiencies)
- generation weights
- Detector efficiencies
- Trigger efficiencies
- Reconstruction efficiencies
- Additional corrections

Each simulated event contributes a certain amount to the final prediction. This contribution is quantified using an event weight. The formula we'll be using is as follow:

$$\text{event weight} = \mathcal{L}\sigma\prod\epsilon_{ff}$$

These factors allow simulation to be converted into a prediction for the number of events expected in the detector.

```python
lumi = 36000.0

def calc_weight(data):
    weight = (
        lumi * data["xsec"] * data["kfac"] * data["filteff"] *
        data["mcWeight"] / data["sum_of_weights"] *
        data["ScaleFactor_PILEUP"] * data["ScaleFactor_FTAG"]
    )
    weight = weight * data["ScaleFactor_MUON"] * data["ScaleFactor_MuTRIGGER"] 
    return weight
```

```python
higgs['eventWeight'] = calc_weight(higgs)
zjets['eventWeight'] = calc_weight(zjets)
diboson['eventWeight'] = calc_weight(diboson)
ttbar['eventWeight'] = calc_weight(ttbar)
```

We can use our understanding of the physics to help us develop a trigger. This trigger will help us choose the "right" leptons. In this analysis, we are only looking for exactly 2 muons, so we keep events where at least one muon fired this lepton trigger. 

```python
def cut_trig(trigM):
    return trigM

data = data[cut_trig(data.trigM)]
higgs = higgs[cut_trig(higgs.trigM)]
zjets = zjets[cut_trig(zjets.trigM)]
diboson = diboson[cut_trig(diboson.trigM)]
ttbar = ttbar[cut_trig(ttbar.trigM)]
```

Now that we know what events fired the trigger, we now want to ensure we have the lepton in every event that fired that trigger. This is matching the lepton to the trigger. Since we could have multiple leptons in every event, we want to have at least one of those leptons matched to the trigger.

```python
def Matched_cut(Matched):
    return ak.sum(Matched == True, axis=1) >= 1

data = data[Matched_cut(data.lep_isTrigMatched)]
higgs = higgs[Matched_cut(higgs.lep_isTrigMatched)]
zjets = zjets[Matched_cut(zjets.lep_isTrigMatched)]
diboson = diboson[Matched_cut(diboson.lep_isTrigMatched)]
ttbar = ttbar[Matched_cut(ttbar.lep_isTrigMatched)]
```

Question 3.30: Are we at a place where can say we found the Higgs?

---

The number of leptons distribution told us that there is very little signal and that we are completely dominated by background. However, the invariant mass distribution tells a different story. The very little signal is predominantly around 125 GeV, which is around the mass of the Higgs boson. 

Now we have to ask ourselves: does this mean we have found the Higgs? How do we know when we have found the Higgs? 

Thus far, we have only looked at simulation, which is all modelled from theory. We expect a contribution around 125 GeV for the Higgs processes because it was modelled like that. Therefore, the only way to truly know is to calculate the ASIMOV statistical significance of our current simulation. 

ASIMOV assumes our real data will have the exact same contribution as our total simulation bin-by-bin, i.e., perfect agreement between data and simulation. To calculate the ASIMOV statistical significance per bin, we use the following formula:

$$Z_i = \sqrt{2\left[\left(S_i+B_i\right)\ln\left(1+\dfrac{S_i}{B_i}\right)\right]}$$

and therefore to calculate to total significance across the distribution:

$$Z_{\text{total}} = \sqrt{\sum_i Z_i^2}$$

Within the field of particle physics, evidence of discovery is when the total significance is $> 3$, but true discovery is when total significance is $> 5$. The language that is used to talk about significance is in terms of sigma, $\sigma$. For example, if the total significance is 5, then it is $5 \sigma$ significance.
